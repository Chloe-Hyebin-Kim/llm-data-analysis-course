## 1. 실행 환경 확인

먼저 현재 Notebook 커널과 작업 폴더를 확인합니다. VS Code에서 프로젝트의 `.venv` 커널을 선택한 뒤 실행하세요.


````python
from pathlib import Path
import sys

print('Python 실행 파일:', sys.executable)
print('현재 작업 폴더:', Path.cwd())
````

## 2. 패키지와 데이터 경로 설정

Notebook의 현재 작업 폴더가 프로젝트 루트인지 `notebooks` 폴더인지에 따라 상대 경로가 달라질 수 있습니다. 아래 코드는 두 경우를 모두 확인해 데이터 폴더를 선택합니다.


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'data' / 'raw').exists() and (PROJECT_ROOT.parent / 'data' / 'raw').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / 'data' / 'raw'
sns.set_theme(style='whitegrid')

print('프로젝트 루트:', PROJECT_ROOT)
print('데이터 폴더:', DATA_DIR)
print('데이터 폴더 존재 여부:', DATA_DIR.exists())
````

## 3. 샘플 데이터 불러오기

`customers.csv`가 없다면 프로젝트 루트의 터미널에서 `python scripts/generate_sample_data.py`를 먼저 실행하세요.


````python
customers_path = DATA_DIR / 'customers.csv'

if not customers_path.exists():
    raise FileNotFoundError(
        f'{customers_path} 파일이 없습니다. '
        '프로젝트 루트에서 python scripts/generate_sample_data.py를 실행하세요.'
    )

customers = pd.read_csv(customers_path)
customers.head()
````

## 4. 기본 구조 확인

행과 열의 개수, 컬럼명, 데이터 타입을 확인합니다.


````python
print('데이터 크기:', customers.shape)
print('컬럼명:', customers.columns.tolist())
customers.info()
````

### 파일 경로 확인 → CSV 읽기 → 데이터 크기 확인 → 컬럼 확인

````python
from pathlib import Path
import pandas as pd

# 데이터 폴더 경로
DATA_DIR = Path("./data")   # 실제 CSV가 있는 폴더에 맞게 수정

# 1. DATA_DIR 존재 여부
print("DATA_DIR 존재 여부:", DATA_DIR.exists())

# 2. customers.csv 존재 여부
customers_path = DATA_DIR / "customers.csv"
print("customers.csv 존재 여부:", customers_path.exists())

# 3. customers.csv 읽기
customers = pd.read_csv(customers_path)

# 4. customers.shape
print("customers.shape:", customers.shape)

# 5. 주요 컬럼
print("주요 컬럼:")
print(customers.columns.tolist())
````
